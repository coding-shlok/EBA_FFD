Go to Altair Files browser, right-click on each folder, copy the full path, and paste below.

```python
# ============================================================================
# PASTE YOUR ABSOLUTE PATHS FROM ALTAIR HERE
# ============================================================================

# Code folder (uploaded separately) — e.g. /data/mpstme-shlocks/EBA-FFD-v2
BASE_PATH = "/data/mpstme-shlocks/EBA-FFD-v2"

# Data folder (uploaded SEPARATELY from code) — e.g. /data/mpstme-shlocks/EBA-FFD-data
DATASET_PATH = "/data/mpstme-shlocks/EBA-FFD-data"

# Preset to run: smoke (fast sanity check) | quick (pilot) | full (reportable)
PRESET = "full"
# ============================================================================

print(f"BASE_PATH: {BASE_PATH}")
print(f"DATASET_PATH: {DATASET_PATH}")
print(f"PRESET: {PRESET}")
```

```python
import os
import sys

# Verify paths BEFORE doing anything else — do not proceed if any check fails
print("Verifying paths...\n")

data_csv = os.path.join(DATASET_PATH, "creditcard.csv")
run_script = os.path.join(BASE_PATH, "run_all_experiments.py")
config_module = os.path.join(BASE_PATH, "config.py")

print(f"BASE_PATH exists: {os.path.isdir(BASE_PATH)}")
print(f"DATASET_PATH exists: {os.path.isdir(DATASET_PATH)}")
print(f"creditcard.csv exists: {os.path.isfile(data_csv)}")
print(f"run_all_experiments.py exists: {os.path.isfile(run_script)}")
print(f"config.py exists: {os.path.isfile(config_module)}")

if not all([
    os.path.isdir(BASE_PATH),
    os.path.isdir(DATASET_PATH),
    os.path.isfile(data_csv),
    os.path.isfile(run_script),
    os.path.isfile(config_module),
]):
    print("\n❌ ERROR: Some paths don't exist. Check the Altair Files tab, "
          "update BASE_PATH / DATASET_PATH in the previous cell, and re-run "
          "both cells before continuing.")
else:
    print("\n✓ All paths valid. Ready to train!")
```

```python
# Install required packages (Altair's base container image doesn't have these)
import subprocess

print("Installing packages...")
pkgs = ["opacus>=1.4.0", "xgboost>=2.0.0", "imbalanced-learn>=0.11.0",
        "shap>=0.43.0", "lime>=0.2.0.1", "seaborn>=0.12.0"]
subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *pkgs])
print("✓ Packages installed.")
```

```python
# Run training with ABSOLUTE PATHS — same pattern as medchain_vit_training:
# a thin notebook launching the real script via subprocess, not inline cells.
import os
import subprocess

os.chdir(BASE_PATH)

# data_loader.py reads this env var to find creditcard.csv in the SEPARATE
# data folder, instead of expecting it nested inside the code folder.
env = os.environ.copy()
env["EBA_FFD_DATA_PATH"] = os.path.join(DATASET_PATH, "creditcard.csv")

cmd = f"python3 {os.path.join(BASE_PATH, 'run_all_experiments.py')} --preset {PRESET}"

print(f"Running command:\n{cmd}\n")
print("=" * 80)

result = subprocess.run(cmd, shell=True, cwd=BASE_PATH, env=env)

print("=" * 80)
if result.returncode == 0:
    print("\n✓ Training completed successfully!")
else:
    print(f"\n❌ Training failed with return code {result.returncode}")
```
